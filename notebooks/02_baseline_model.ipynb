# Baseline Complex-Systems Model

This notebook runs the unconstrained baseline model and inspects key trajectory statistics.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == "notebooks" else cwd
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

repo_root
```

```python
import numpy as np
import matplotlib.pyplot as plt

from experiments.models.baseline import BaselineComplexSystemModel
from experiments.models import run_simulation
from experiments.core import DistinctionSet
```

```python
config = {
    "n_agents": 20,
    "n_features": 3,
    "interaction_strength": 0.3,
    "steps": 100,
    "seed": 42,
}

model = BaselineComplexSystemModel(
    n_agents=config["n_agents"],
    n_features=config["n_features"],
    interaction_strength=config["interaction_strength"],
    seed=config["seed"],
)

result = run_simulation(model, steps=config["steps"], distinctions=DistinctionSet([]))
trajectory = np.array([state.state for state in result.trajectory])

print("Trajectory array shape:", trajectory.shape)
print("Mean state value:", round(float(trajectory.mean()), 4))
print("Std state value:", round(float(trajectory.std()), 4))
```

```python
mean_over_time = trajectory.mean(axis=1)
var_over_time = trajectory.var(axis=1)

divergence_over_time = [
    np.linalg.norm(trajectory[t] - trajectory[0])
    for t in range(len(trajectory))
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(mean_over_time)
axes[0].set_title("Mean State")
axes[0].set_xlabel("Step")

axes[1].plot(var_over_time, color="orange")
axes[1].set_title("State Variance")
axes[1].set_xlabel("Step")

axes[2].plot(divergence_over_time, color="red")
axes[2].set_title("Divergence from Initial State")
axes[2].set_xlabel("Step")

for ax in axes:
    ax.grid(alpha=0.3)

plt.tight_layout()
```
